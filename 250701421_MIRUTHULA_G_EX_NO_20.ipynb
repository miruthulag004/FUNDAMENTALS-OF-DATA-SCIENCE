```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans

# Generate a sample Mall Customers dataset with 200 records
np.random.seed(42)

df = pd.DataFrame({
    'CustomerID': range(1, 201),
    'Gender': np.random.choice(['Male', 'Female'], 200),
    'Age': np.random.randint(18, 71, 200),
    'Annual Income (k$)': np.random.randint(15, 138, 200),
    'Spending Score (1-100)': np.random.randint(1, 101, 200)
})

# Display dataset information
df.info()
print(df.head())

# Pairplot
sns.pairplot(df)
plt.show()

# Select Annual Income and Spending Score
features = df.iloc[:, [3, 4]].values

# Apply K-Means clustering
model = KMeans(n_clusters=5, random_state=42, n_init=10)
model.fit(features)

# Add cluster labels
Final = df.iloc[:, [3, 4]].copy()
Final['label'] = model.labels_

print(Final.head())

# Visualize clusters
sns.set_style("whitegrid")
sns.FacetGrid(Final, hue="label", height=6) \
    .map(plt.scatter, "Annual Income (k$)",
         "Spending Score (1-100)") \
    .add_legend()

plt.show()

# Elbow method
features_el = df.iloc[:, [2, 3, 4]].values
wcss = []

for i in range(1, 10):
    model = KMeans(n_clusters=i, random_state=42, n_init=10)
    model.fit(features_el)
    wcss.append(model.inertia_)

plt.plot(range(1, 10), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```
