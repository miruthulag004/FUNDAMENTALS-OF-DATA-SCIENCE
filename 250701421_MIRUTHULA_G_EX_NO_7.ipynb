```python

# Experiment 7: Random Sampling and Sampling Distribution
# Objective:
# To explore random sampling and understand sampling distributions
# using Python.
# Step 1: Import required libraries
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducible results
np.random.seed(42)

# Step 2: Generate a population
population_mean = 50
population_std = 10
population_size = 100000

population = np.random.normal(
    population_mean,
    population_std,
    population_size
)

print("Population Mean:", np.mean(population))
print("Population Standard Deviation:", np.std(population))

# Plot the population distribution
plt.figure(figsize=(10, 5))
plt.hist(population, bins=50, color='skyblue', edgecolor='black',
         density=True, alpha=0.7)
plt.axvline(np.mean(population), color='red', linestyle='dashed',
            linewidth=2, label='Population Mean')
plt.title("Population Distribution")
plt.xlabel("Values")
plt.ylabel("Density")
plt.legend()
plt.show()

# Step 3: Perform random sampling
sample_sizes = [30, 50, 100]
num_samples = 1000

sample_means = {}
sample_stds = {}

for size in sample_sizes:
    sample_means[size] = []
    sample_stds[size] = []

    for _ in range(num_samples):
        sample = np.random.choice(
            population,
            size=size,
            replace=False
        )

        sample_means[size].append(np.mean(sample))
        sample_stds[size].append(np.std(sample, ddof=1))

# Step 4: Display sample statistics
for size in sample_sizes:
    print(f"\nSample Size: {size}")
    print("Mean of Sample Means:", np.mean(sample_means[size]))
    print("Standard Deviation of Sample Means:", np.std(sample_means[size]))
    print("Average Sample Standard Deviation:", np.mean(sample_stds[size]))

# Step 5: Plot sampling distributions
plt.figure(figsize=(12, 10))

for i, size in enumerate(sample_sizes):
    plt.subplot(len(sample_sizes), 1, i + 1)

    plt.hist(
        sample_means[size],
        bins=30,
        color='lightgreen',
        edgecolor='black',
        alpha=0.7
    )

    plt.axvline(
        np.mean(population),
        color='red',
        linestyle='dashed',
        linewidth=2,
        label='Population Mean'
    )

    plt.title(f"Sampling Distribution of Sample Means (n={size})")
    plt.xlabel("Sample Mean")
    plt.ylabel("Frequency")
    plt.legend()

plt.tight_layout()
plt.show()

# Step 6: Demonstrate the Central Limit Theorem
# Compare sampling distributions for different sample sizes.

plt.figure(figsize=(10, 6))

for size in sample_sizes:
    plt.hist(
        sample_means[size],
        bins=30,
        density=True,
        alpha=0.4,
        label=f"Sample Size {size}"
    )

plt.axvline(
    np.mean(population),
    color='red',
    linestyle='dashed',
    linewidth=2,
    label='Population Mean'
)

plt.title("Central Limit Theorem: Sampling Distributions")
plt.xlabel("Sample Mean")
plt.ylabel("Density")
plt.legend()
plt.show()

print("\nExperiment 7 completed successfully!")
```
