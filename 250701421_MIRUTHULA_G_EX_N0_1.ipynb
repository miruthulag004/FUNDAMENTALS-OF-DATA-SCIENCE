```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Year":[2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023],
    "Job_postings": [5000, 6500, 8000, 9500, 12000, 15000, 18000, 22000, 26000, 30000]
}
df = pd.DataFrame(data)
print(df)
plt.figure(figsize=(5,5))
plt.plot(df['Year'], df['Job_postings'], marker ='o', color = 'pink', linewidth = 2)
plt.title("Trend of data science jobs")
plt.xlabel("Year")
plt.ylabel("Number of job postings")
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "role": [
        "data analyst", "data scientist", "data engineer",
        "data analyst", "machine learning engineer",
        "data scientist", "data engineer", "data analyst",
        "business analyst", "data scientist",
        "data engineer", "business analyst"
    ]
}

df = pd.DataFrame(data)

role_counts = df["role"].value_counts()

print(role_counts)

plt.figure(figsize=(8, 5))
role_counts.plot(kind="bar", color="violet", edgecolor="black")
plt.title("distribution of data science roles")
plt.xlabel("job role")
plt.ylabel("number of employees")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.show()
```

```python
from cryptography.fernet import Fernet

key = Fernet.generate_key()
cipher = Fernet(key)

data = "My password is Secure@123"

print("Original Data:")
print(data)

encrypted_data = cipher.encrypt(data.encode())

print("\nEncrypted Data:")
print(encrypted_data.decode())

decrypted_data = cipher.decrypt(encrypted_data).decode()

print("\nDecrypted Data:")
print(decrypted_data)
```
