```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_excel('/content/Social_Network_Ads.xlsx')

# Display the dataset
data

# Dataset information
data.info()

# Summary statistics
data.describe()

# Count values of Gender
data.value_counts('Gender')

# Count plot
sns.countplot(x='Gender', data=data)
plt.show()

# Convert Gender into dummy variables
dummies = pd.get_dummies(data.Gender)

FinalDataset = pd.concat(
    [dummies, data.iloc[:, [0, 2, 3, 4]]],
    axis=1
)

# Display final dataset
FinalDataset.head()

# Scatter plot: Age vs Estimated Salary
sns.scatterplot(
    x='Age',
    y='EstimatedSalary',
    hue='Gender',
    data=data
)
plt.show()

# Scatter plot: Age vs Estimated Salary based on Purchase
sns.scatterplot(
    x='Age',
    y='EstimatedSalary',
    hue='Purchased',
    data=data
)
plt.show()

# Pairplot
g = sns.pairplot(
    data[['Age', 'EstimatedSalary', 'Purchased']],
    hue='Purchased',
    height=3
)

# Correctly rotate x-axis labels on all subplots
for ax in g.axes.flat:
    if ax is not None:
        ax.tick_params(axis='x', labelrotation=45)

plt.show()

# Histogram of Estimated Salary
sns.FacetGrid(
    data,
    hue='Gender',
    height=5
).map(
    sns.histplot,
    'EstimatedSalary'
).add_legend()

plt.show()

# Histogram of Age
sns.FacetGrid(
    data,
    hue='Gender',
    height=5
).map(
    sns.histplot,
    'Age'
).add_legend()

plt.show()
```
